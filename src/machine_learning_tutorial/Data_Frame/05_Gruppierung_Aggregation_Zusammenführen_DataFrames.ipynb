## Von Zeilen zu Antworten

- Eine Tabelle hat **viele Zeilen** — aber wir wollen meist eine **kurze Antwort**
- "Gesamtumsatz **pro Kategorie**?"
- "Durchschnittlicher Bestellwert **pro Region**?"
- Wir müssen **Gruppen von Zeilen** zu einzelnen Zahlen **zusammenfassen**
- pandas erledigt das mit **Gruppierung** und **Aggregation**

```python
import pandas as pd
```


## Unser Beispiel: Eine kleine Bestellungstabelle

- Jede Zeile ist eine **Bestellung**
- Spalten: `category`, `region`, `amount` (in Euro), `units`
- Klein genug, um die Ergebnisse von Hand zu überprüfen

```python
orders = pd.DataFrame({
    "category": ["Books", "Books", "Toys", "Toys", "Books",
                 "Toys", "Books", "Toys", "Books", "Toys"],
    "region":   ["North", "South", "North", "South", "North",
                 "North", "South", "South", "North", "South"],
    "amount":   [30, 45, 60, 20, 25, 80, 50, 35, 40, 70],
    "units":    [3, 4, 2, 1, 2, 3, 5, 1, 4, 2],
})
```

```python
orders
```


## Die Idee: Split – Apply – Combine

Gruppierung funktioniert immer in drei Schritten:

- **Split**: Zeilen in Gruppen aufteilen (z. B. eine Gruppe pro Kategorie)
- **Apply**: für jede Gruppe etwas berechnen (z. B. die Summe)
- **Combine**: die Ergebnisse in einer neuen, kleinen Tabelle zusammenfassen


## groupby: Das grundlegende Muster

- `df.groupby("category")` teilt die Zeilen nach Kategorie auf
- Dann eine Spalte auswählen: `["amount"]`
- Dann eine Zusammenfassung anwenden: `.sum()`
- Lesen Sie es als: **Summe von amount, pro Kategorie**

```python
orders.groupby("category")["amount"].sum()
```


## Andere Zusammenfassungen

- `.mean()` — der Durchschnitt
- `.count()` — wie viele Zeilen in jeder Gruppe
- `.max()` / `.min()` — größter / kleinster Wert
- Gleiches Muster, nur ein anderes Wort am Ende

```python
orders.groupby("category")["amount"].mean()
```

```python
orders.groupby("category")["amount"].count()
```


## Mehrere Zusammenfassungen auf einmal

- Oft möchten wir Summe **und** Mittelwert **und** Anzahl zusammen
- Übergeben Sie eine **Liste** von Zusammenfassungen an `.agg([...])`
- Das Ergebnis hat eine Spalte pro Zusammenfassung

```python
orders.groupby("category")["amount"].agg(["sum", "mean", "count"])
```


## Verschiedene Zusammenfassung pro Spalte

- Verwenden Sie ein **Dictionary**: Spalte → Zusammenfassung
- Hier: Summe von `amount`, aber Durchschnitt von `units`

```python
orders.groupby("category").agg({"amount": "sum", "units": "mean"})
```


## Gruppierung nach mehreren Spalten

- Übergeben Sie eine **Liste** von Spalten an `groupby`
- Hier: eine Gruppe pro Kombination aus `category` **und** `region`
- Das Ergebnis hat eine Zeile für jede Kombination
- Sein Index hat jetzt **zwei Ebenen** (`category`, dann `region`)

```python
orders.groupby(["category", "region"])["amount"].sum()
```


## Kategorien zählen: value_counts

- Eine schnelle Möglichkeit zu zählen, wie oft jeder Wert vorkommt
- Funktioniert direkt auf einer einzelnen Spalte
- Ideal für Text-/Kategoriespalten

```python
orders["region"].value_counts()
```


## pivot_table: Eine kompakte Kreuztabelle

- Wandelt eine Gruppierungsspalte in **Zeilen** um, eine andere in **Spalten**
- `values`: die Zahlen, die zusammengefasst werden sollen
- `aggfunc`: wie sie zusammengefasst werden (hier: `"sum"`)
- `margins=True` ergänzt eine zusätzliche `All`-Zeile und -Spalte mit den Summen

```python
orders.pivot_table(values="amount", index="category",
                   columns="region", aggfunc="sum")
```


## DataFrames kombinieren

- Daten sind oft auf **mehrere Tabellen** verteilt
- Wir müssen sie zusammenführen, bevor wir sie analysieren
- Zwei Hauptmethoden:
  - **Stacking**: Tabellen übereinanderlegen (`pd.concat`)
  - **Joining**: Zeilen über einen gemeinsamen Schlüssel verknüpfen (`pd.merge`)


## Stapeln mit concat

- Gleiche Spalten, mehr Zeilen (z. B. Januar- und Februar-Bestellungen)
- `pd.concat([df1, df2])` stapelt sie übereinander

```python
jan = pd.DataFrame({"category": ["Books", "Toys"], "amount": [30, 60]})
feb = pd.DataFrame({"category": ["Books", "Toys"], "amount": [45, 20]})
```

```python
pd.concat([jan, feb], ignore_index=True)
```


## Joining mit merge

- Zwei Tabellen teilen eine **Key**-Spalte (z. B. `category`)
- `pd.merge` verbindet Zeilen mit demselben Key
- Hier: eine Beschreibung aus einer **Lookup-Tabelle** hinzufügen

```python
category_info = pd.DataFrame({
    "category": ["Books", "Toys"],
    "department": ["Media", "Children"],
})
```

```python
pd.merge(orders, category_info, on="category").head()
```


## Inner Join vs. Left Join

- **Inner** (Standard): nur Zeilen behalten, die in **beiden** Tabellen übereinstimmen
- **Left**: **alle** Zeilen der linken Tabelle behalten, Lücken mit `NaN` füllen
- Wählen Sie `how="inner"` oder `how="left"`

```python
left = pd.DataFrame({"category": ["Books", "Toys", "Games"], "amount": [30, 60, 10]})
lookup = pd.DataFrame({"category": ["Books", "Toys"], "department": ["Media", "Children"]})
```

```python
pd.merge(left, lookup, on="category", how="inner")
```

```python
pd.merge(left, lookup, on="category", how="left")
```


## Häufige Fehler beim Joining

- **Nicht übereinstimmende Keys**: `"Books"` vs `"books"` werden nicht gefunden (vorher bereinigen!)
- **Fehlende Keys**: ein Inner Join **verwirft** nicht übereinstimmende Zeilen stillschweigend
- **Doppelte Keys**: ein Key, der zweimal vorkommt, kann Zeilen **vervielfachen**
- Überprüfen Sie die Zeilenanzahl **vor und nach** einem Merge


## Alles zusammenführen

- Zuerst **gruppieren**, um Summen pro Kategorie zu erhalten
- `reset_index()` macht die Gruppen-Labels wieder zu einer normalen Spalte, damit wir darauf `merge`n können
- Dann eine **Lookup-Tabelle** mit dem Ergebnis **zusammenführen**
- Dies ist ein sehr verbreiteter Arbeitsablauf in der Praxis

```python
totals = orders.groupby("category")["amount"].sum().reset_index()
```

```python
totals
```

```python
pd.merge(totals, category_info, on="category")
```


## Zusammenfassung

- **groupby** = Split → Apply → Combine (aufteilen → anwenden → zusammenführen)
- `.sum()`, `.mean()`, `.count()`; mehrere auf einmal mit `.agg([...])`
- Nach **mehreren Spalten** gruppieren und Kreuztabellen mit `pivot_table` erstellen
- **concat** stapelt Tabellen; **merge** verbindet sie über einen Schlüssel
- Achten Sie auf nicht übereinstimmende Schlüssel und doppelte Zeilen


## Workshop: Gruppieren und kombinieren

Ein Café erfasst eine Zeile pro Bestellung: welches `product`, in welcher
`shift`, den `sales` in Euro und wie viele `items`. Verwenden Sie die
Tabelle `orders` sowie die Lookup- und Monatstabellen unten:

1. Gesamter `sales` **pro Schicht**
2. Durchschnittliche `items` **pro Produkt**
3. Gesamter `sales` **pro Produkt _und_ Schicht** (eine **Liste** in `groupby`)
4. Bestellungen **pro Produkt** (mit `value_counts`)
5. Summe **und** Anzahl von `sales` pro Produkt (ein `.agg`, eine **Liste**)
6. Gesamter `sales`, aber **Durchschnitt** der `items` pro Produkt (ein `.agg`, ein **Dictionary**)
7. **Stapeln** Sie die Januar- und Februar-Tabelle mit `pd.concat`
8. Fügen Sie `shift_info` per Merge zu `orders` hinzu (Schlüssel: `shift`)
9. Gesamter `sales` pro Schicht, dann `shift_info` per Merge auf das Ergebnis anwenden
10. Verbinden Sie die gestapelte Monatstabelle mit `product_info` (Schlüssel: `product`)
    mit `how="inner"`, dann `how="left"` — welches Produkt lässt inner weg?
11. **Bonus (optional)**: `pivot_table` des gesamten `sales`, `product` als Zeilen,
    `shift` als Spalten, **mit Zeilen-/Spaltensummen** (`margins=True`)

```python
orders = pd.DataFrame({
    "product": ["Coffee", "Coffee", "Cake", "Cake", "Coffee",
                "Cake", "Coffee", "Cake", "Coffee", "Cake"],
    "shift":   ["Morning", "Afternoon", "Morning", "Afternoon", "Morning",
                "Morning", "Afternoon", "Afternoon", "Morning", "Afternoon"],
    "sales":   [30, 45, 60, 20, 25, 80, 50, 35, 40, 70],
    "items":   [3, 4, 2, 1, 2, 3, 5, 1, 4, 2],
})
```

```python
shift_info = pd.DataFrame({
    "shift": ["Morning", "Afternoon"],
    "barista": ["Alice", "Bob"],
})
```

```python
product_info = pd.DataFrame({
    "product": ["Coffee", "Cake"],
    "station": ["Barista", "Kitchen"],
})
```

```python
jan_orders = pd.DataFrame({"product": ["Coffee", "Cake"], "sales": [30, 60]})
feb_orders = pd.DataFrame({"product": ["Coffee", "Juice"], "sales": [45, 15]})
```

*Antwort:* 
### Lösung

```python
orders.groupby("shift")["sales"].sum()
```

```python
orders.groupby("product")["items"].mean()
```

```python
orders.groupby(["product", "shift"])["sales"].sum()
```

```python
orders["product"].value_counts()
```

```python
orders.groupby("product")["sales"].agg(["sum", "count"])
```

```python
orders.groupby("product").agg({"sales": "sum", "items": "mean"})
```

```python
pd.concat([jan_orders, feb_orders], ignore_index=True)
```

```python
pd.merge(orders, shift_info, on="shift").head()
```

```python
shift_totals = orders.groupby("shift")["sales"].sum().reset_index()
pd.merge(shift_totals, shift_info, on="shift")
```

```python
monthly = pd.concat([jan_orders, feb_orders], ignore_index=True)
```

```python
pd.merge(monthly, product_info, on="product", how="inner")
```

```python
pd.merge(monthly, product_info, on="product", how="left")
```


- Der Inner Join lässt **Juice** weg (keine Zeile in `product_info`); der Left
  Join behält es mit fehlender `station` (`NaN`)

```python
orders.pivot_table(values="sales", index="product",
                   columns="shift", aggfunc="sum", margins=True)
```


- `margins=True` ergänzt eine `All`-Zeile und eine `All`-Spalte mit den
  Summen — dieselben Zahlen wie aus separaten `groupby`-Summen, in einer
  Tabelle
