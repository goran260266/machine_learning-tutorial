## Warum saubere Daten?

- Echte Daten sind **unordentlich**: Lücken, Duplikate, Tippfehler, Extremwerte
- Modelle und Statistiken sind nur so gut wie die Daten
- "Garbage in, garbage out"
- Bereinigung ist oft der **größte Teil** eines Datenprojekts

```python
import pandas as pd
import numpy as np
```


## Ein unordentlicher Datensatz

- Fehlende Werte, eine doppelte Zeile, ein extremer Wert, inkonsistenter Text
- Klein genug, um jede Korrektur manuell zu überprüfen

```python
df = pd.DataFrame({
    "name": ["Alice", "Bob", "Bob", "Dana", "Eve"],
    "age": [25, 30, 30, np.nan, 250],
    "city": ["Berlin", "munich ", "munich ", "Berlin", "HAMBURG"],
    "spend": [100.0, 200.0, 200.0, 150.0, np.nan],
})
```

```python
df
```


## Schritt 1: Fehlende Werte finden

- Fehlende Werte erscheinen als `NaN` (*Not a Number* — die Markierung von pandas für eine Lücke)
- `isnull()` markiert jede Lücke als `True`
- `.sum()` zählt Lücken pro Spalte

```python
df.isnull().sum()
```

```python
df["age"].isnull().sum()
```


## Warum Werte fehlen — und warum das wichtig ist

- Häufige Ursachen: nicht erfasst, ein optionales Feld, ein Sensor- oder Importfehler
- Eine Lücke kann **jede spätere Zusammenfassung verzerren** (ein Mittelwert über fehlende Daten ist irreführend)
- Es gibt keine einheitlich richtige Lösung — **entscheiden Sie spaltenweise**, was eine Lücke bedeutet
- Zwei Hauptoptionen: **Zeile entfernen** oder **Lücke füllen**


## Lücken behandeln: Entfernen

- `df.dropna()` entfernt **jede Zeile**, die eine Lücke enthält
- Einfach — aber Sie verlieren ganze Zeilen, einschließlich ihrer guten Werte
- Sinnvoll, wenn Lücken **selten** sind; verschwenderisch, wenn sie häufig sind

```python
df.shape
```

```python
df
```

```python
df_ = df.dropna()
```

```python
df_
```


## Lücken behandeln: Füllen

- `df.fillna(value)` ersetzt jede Lücke durch einen Wert
- Bei **Zahlen** ist der **Median** eine sichere Wahl (robust gegenüber Ausreißern)
- Bei **Kategorien** ist der häufigste Wert eine gängige Wahl

```python
df["spend"].fillna(df["spend"].median())
```


## pandas gibt ein **neues** Ergebnis zurück – weisen Sie es wieder zu!

- Ein Bereinigungsschritt verändert `df` **nicht** direkt
- Er liefert Ihnen ein **neues** Ergebnis – Sie müssen es **wieder zuweisen**, um es zu behalten
- Andernfalls wird die Änderung einmal angezeigt und dann verworfen
- Dies ist der mit Abstand häufigste Anfängerfehler – achten Sie bei jedem Schritt darauf

```python
df["spend"] = df["spend"].fillna(df["spend"].median())
```

```python
df
```


## Schritt 2: Doppelte Zeilen

- `duplicated()` markiert Zeilen, die eine frühere Zeile wiederholen
- `drop_duplicates()` entfernt sie
- Prüfen Sie, ob ein Duplikat **echt** ist (zwei verschiedene Personen) oder ein **Artefakt**
  (derselbe Datensatz wurde zweimal importiert)

```python
df.duplicated().sum()
```

```python
df = df.drop_duplicates()
```

```python
df
```


## Schritt 3: Ausreißer

- Ausreißer sind unrealistische oder extreme Werte (age = 250!)
- Erkennen Sie sie mit `describe()` — schauen Sie sich `min` / `max` an
- Dann entscheiden Sie: **korrigieren**, die Zeile **löschen** oder **begrenzen** (`clip`)
- Hinweis: `df[df["age"] < 120]` löscht auch Zeilen, in denen `age` fehlt
  (`NaN < 120` ist `False`) — hier praktisch, aber beachten Sie das

```python
df.describe
```

```python
df["age"].describe()
```

```python
df = df[df["age"] < 120]
```

```python
df
```


## Schritt 4: Inkonsistenter Text

- "munich ", "HAMBURG", "Berlin" — unterschiedliche Schreibweisen und überflüssige Leerzeichen
- `.str` wendet String-Methoden (`strip`, `title`, …) auf eine ganze Spalte auf einmal an
- `strip()` entfernt Leerzeichen; `title()` vereinheitlicht die Großschreibung
- **Warum es wichtig ist:** Inkonsistenter Text führt dazu, dass `groupby` / `merge` später falsche Ergebnisse liefert —
  `"munich "` und `"Munich"` werden als zwei verschiedene Gruppen gezählt

```python
df["city"] = df["city"].str.strip().str.title()
```

```python
df
```


## Schritt 5: Datentypen

- Zahlen, die als **Text** gespeichert sind, können nicht berechnet werden
- Typen explizit konvertieren:
  - `df["age"].astype(int)` — erst **nach** dem Füllen fehlender Werte
    (`astype(int)` schlägt fehl bei einer Spalte, die noch `NaN` enthält)
  - `pd.to_datetime(df["date"])` wandelt Text in echte Datumsangaben um —
    beides machen wir gleich an einer echten Datei


Ein kurzes Beispiel — Text wird in ganze Zahlen umgewandelt:

```python
pd.Series(["1", "2", "3"]).astype(int)
```


## Das Chaos kommt meist in einer Datei

- Bisher haben wir eine selbst gebaute Tabelle bereinigt
- In der Praxis kommt das Chaos in einer **CSV-Datei**, die Ihnen jemand übergibt
- Hier ein kleiner Auszug aus einem Bestell-Export — sehen Sie sich den **Rohtext** an:
  `?` für einen unbekannten Preis, eine leere Zelle, das Wort `unknown`, zwei Datumsformate

```python
from pathlib import Path

csv_text = """order_date,category,price,qty
2026-03-05,Electronics,199.99,1
2026/03/07,Electronics,?,2
2026-03-08,,24.50,1
2026/03/12,Clothing,49.90,3
2026-03-14,unknown,12.99,2
2026-03-19,Sports,89.00,1
"""
Path("orders_sample.csv").write_text(csv_text, encoding="utf-8")
```


## Ein naiver Lade-Versuch

- `read_csv` liest die Datei ohne Beschwerde — aber prüfen Sie `dtypes`!
- `price` kam als **Text** an: das `?` hat pandas daran gehindert, Zahlen zu erkennen
- `category` enthält noch das **Wort** `unknown` — pandas weiß nicht,
  dass es "fehlend" bedeutet

```python
orders = pd.read_csv("orders_sample.csv")
```

```python
orders
```

```python
orders.dtypes
```


## Die Markierungen benennen: `na_values`

- Gängige Lücken-Markierungen erkennt pandas von selbst (leere Zelle, `NA`, `n/a`)
- **Eigene** Markierungen wie `?` oder `unknown` müssen Sie über `na_values` benennen
- Jetzt ist `price` eine echte Zahlenspalte, und jede Lücke ist ein echtes `NaN`

```python
orders = pd.read_csv("orders_sample.csv", na_values=["?", "unknown"])
```

```python
orders
```

```python
orders.dtypes
```


## Text-Datumsangaben → echte Datumsangaben: `to_datetime`

- `order_date` ist noch **Text** — damit können Sie nicht rechnen
- Unsere Datei mischt zwei Datumsformate, deshalb verweigert das einfache
  `pd.to_datetime` — die Fehlermeldung nennt die Lösung: `format="mixed"`
- **Allgemeine Vorsicht:** Wenn Datumsangaben *möglicherweise* tageszuerst sind
  (in europäischen Dateien üblich), prüfen Sie nach dem Konvertieren `min()` / `max()`
  — ein stilles Vertauschen von Tag und Monat übersieht man leicht

```python
# pd.to_datetime(orders["order_date"])  # ValueError: two different formats!
```

```python
orders["order_date"] = pd.to_datetime(orders["order_date"], format="mixed")
```

```python
orders.dtypes
```


Mit echten Datumsangaben kann man rechnen — z. B. den Monat extrahieren:

```python
orders["order_date"].dt.month
```


## Eine Kategorie-Lücke füllen: der häufigste Wert

- Der Median ergibt für **Text**-Spalten keinen Sinn
- Verwenden Sie stattdessen den häufigsten Wert: `mode()[0]`
- Wie immer: **Ergebnis wieder zuweisen**

```python
orders["category"].mode()[0]
```

```python
orders["category"] = orders["category"].fillna(orders["category"].mode()[0])
```

```python
orders["price"] = orders["price"].fillna(orders["price"].median())
```


## Spalten umbenennen, Spalten erzeugen

- Kryptische Spaltennamen verwirren später alle — `rename` behebt das
- Eine **neue Spalte** aus vorhandenen: einfach einem noch nicht existierenden
  Namen zuweisen
- Beides sind Alltagsschritte beim Aufräumen eines frischen Datensatzes

```python
orders = orders.rename(columns={"qty": "quantity"})
```

```python
orders["revenue"] = orders["price"] * orders["quantity"]
```

```python
orders
```


## Zusammenfassung

- Bereinigen: fehlende Werte, Duplikate, Ausreißer, inkonsistenter Text, Typen
- Lücken **löschen oder füllen**; Duplikate entfernen; `min`/`max` auf Ausreißer prüfen
- Unordentliche Dateien laden: `na_values` für Lücken-Markierungen, `to_datetime`
  für Datumsangaben, `rename` und neue Spalten zum Aufräumen
- **Ergebnis immer wieder zuweisen** — pandas gibt ein neues Objekt zurück
- Zuerst bereinigen — alles Nachfolgende hängt davon ab

**Weiter**: bereinigte Daten für Modelle vorbereiten

## Mini-Workshop: eine unordentliche Datei

Die Zelle unten schreibt eine kleine unordentliche CSV-Datei. Laden und
bereinigen Sie sie:

1. Laden Sie `sales_sample.csv` so, dass `?` und `unknown` echte Lücken
   werden — prüfen Sie `dtypes`: Ist `price` jetzt eine Zahlenspalte?
2. Wandeln Sie `day` in echte Datumsangaben um (zwei Datumsformate!)
3. Füllen Sie die Lücke in `product` mit dem **häufigsten** Produkt und die
   Lücke in `price` mit dem **Median**-Preis
4. Benennen Sie `day` → `order_date` und `qty` → `quantity` um
5. Erzeugen Sie eine Spalte `revenue` (`price * quantity`) — wie hoch ist
   der **Gesamtumsatz**?

```python
sales_text = """day,product,price,qty
2026-04-02,Pen,1.50,10
2026/04/03,Notebook,?,3
2026-04-05,Pen,1.50,5
2026/04/07,unknown,4.20,2
2026-04-09,Marker,2.80,4
2026-04-10,Pen,3.90,1
"""
Path("sales_sample.csv").write_text(sales_text, encoding="utf-8")
```

*Antwort:* 
### Lösung

```python
sales = pd.read_csv("sales_sample.csv", na_values=["?", "unknown"])
```

```python
sales.dtypes
```

```python
sales["day"] = pd.to_datetime(sales["day"], format="mixed")
```

```python
sales["product"] = sales["product"].fillna(sales["product"].mode()[0])
```

```python
sales["price"] = sales["price"].fillna(sales["price"].median())
```

```python
sales = sales.rename(columns={"day": "order_date", "qty": "quantity"})
```

```python
sales["revenue"] = sales["price"] * sales["quantity"]
```
