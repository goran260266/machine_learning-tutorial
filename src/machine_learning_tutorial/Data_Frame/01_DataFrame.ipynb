```python
# Example of customer data frame with lognormal income and random purchases
n = 10
customers = pd.DataFrame({
    "customer_id": np.arange(1, n + 1),
    "age": rng.integers(18, 70, n),
    "region": rng.choice(["North", "South", "East", "West"], n),
    "income": rng.lognormal(mean=8.0, sigma=0.4, size=n).round(0),
    "purchases": rng.integers(0, 25, n),
})
customers
```

### Neue Zeile mit in DataFrame Einfogen

```python
# Option 1: Über die .loc-Zuweisung (Schnell und einfach)
# Wenn Sie den Index direkt auf die Länge des DataFrames setzen, wird die Zeile automatisch am Ende angehängt.
# Eine neue Zeile am Ende einfügen
customers.loc[len(customers)] = {
    "customer_id": 11,
    "age": 45,
    "region": "West",
    "income": 1200000,  # Der Millionär
    "purchases": 50
}

customers.tail()

```

```python

# Option 2: Mit pd.concat() (Empfohlener Pandas-Standard)
# Das Erstellen eines neuen DataFrames für die neue Zeile und das anschließende Verknüpfen ist besonders sauber, da append() in neueren Pandas-Versionen entfernt wurde.
# Neuen Kunden als DataFrame definieren
new_customer = pd.DataFrame([{
    "customer_id": 11,
    "age": 38,
    "region": "North",
    "income": 2500000,  # Der Millionär
    "purchases": 12
}])

# Am Ende des bestehenden DataFrames anhängen
customers = pd.concat([customers, new_customer], ignore_index=True)
customers.tail()
```

```python
# 1. Erstellen Sie die neue Spalte als Series (oder DataFrame) mit passender Länge und concatieren
spend_column = pd.Series(
    (customers["purchases"] * rng.normal(40, 8, len(customers))).round(2).clip(lower=0),
    name="spend"
)

# 2. Mit axis=1 nebeneinander an das bestehende DataFrame anheften
customers = pd.concat([customers, spend_column], axis=1)
customers
```

```python
# Add spend column to customers data frame
customers["spend"] = (customers["purchases"] * rng.normal(40, 8, len(customers))).round(2).clip(lower=0)
customers
```

```python
customers.to_csv("shop_customers.csv", index=False)
read = pd.read_csv("shop_customers.csv")
read
```
