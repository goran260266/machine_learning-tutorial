```python
import numpy as np
import pandas as pd

# 1. Hier sind die echten numerischen Werte: [20000, 35000, 40000, 45000, 60000]
data = {"Einkommen":[20000, 35000, 40000, 45000, 60000]}
df = pd.DataFrame(data)

# 2. Mittelwert (mu) und Standardabweichung (sigma) berechnen
mu = df["Einkommen"].mean()  # Ergibt exakt: 40000.0
sigma = df["Einkommen"].std(
    ddof=0
)  # Ergibt exakt: 12649.11 (Populations-Standardabweichung)

# 3. Die Z-Score-Formel anwenden: (x - mu) / sigma
df["Einkommen_skaliert"] = (df["Einkommen"] - mu) / sigma

# Ergebnisse runden für eine schönere Anzeige
df["Einkommen_skaliert"] = df["Einkommen_skaliert"].round(2)

print("Mittelwert (mu):", mu)
print("Standardabweichung (sigma):", round(sigma, 2))
print("\nDer transformierte Datensatz:")
print(df)
```


2. Der professionelle Weg im ML (Mit Scikit-Learn)

In der Praxis tippst du die Formel nicht selbst ein. Du nutzt den StandardScaler von Scikit-Learn. Dieser merkt sich die Werte aus den Trainingsdaten, um sie später auch auf Testdaten anzuwenden.

```python
from sklearn.preprocessing import StandardScaler

# Daten vorbereiten (Scikit-Learn erwartet ein 2D-Array/DataFrame)
X_train = df[["Einkommen"]]

# 1. Den Scaler initialisieren
scaler = StandardScaler()

# 2. "fit" berechnet mu und sigma | "transform" wendet die Formel an
X_train_scaled = scaler.fit_transform(X_train)

# Das Ergebnis ist ein NumPy-Array mit exakt denselben skalierten Werten:
print(X_train_scaled)
```
