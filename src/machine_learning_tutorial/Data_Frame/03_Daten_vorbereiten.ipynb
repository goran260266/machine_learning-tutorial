## Von sauberen Daten zu Modell-Eingaben

- Bereinigung hat Fehler entfernt — aber Modelle brauchen mehr als saubere Daten
- Sie benötigen **nur Zahlen** und **vergleichbare Wertebereiche**
- Dieser Schritt wird **Feature-Vorbereitung** genannt
- Das Ergebnis ist die Eingabe für die Machine-Learning-Modelle später in dieser Woche

```python
import pandas as pd
```

```python
people = pd.DataFrame({
    "age": [25, 40, 60, 35],
    "income": [2000, 5000, 9000, 3000],
    "region": ["North", "South", "North", "West"],
})
```

```python
people
```


## Warum Skalierung?

- Spalten können **sehr unterschiedliche Wertebereiche** haben (Alter 18–70, Einkommen in Tausend)
- Viele Modelle lassen die Spalte mit dem **großen Wertebereich** die kleine dominieren
- **Skalierung** bringt jede Spalte in einen vergleichbaren Bereich
- Dann trägt jedes Feature gleichberechtigt bei


## `StandardScaler` anwenden

- `StandardScaler` aus scikit-learn skaliert jede Spalte neu
- `fit_transform` erlernt die Skalierung und wendet sie in einem Schritt an
- Das Ergebnis hat **Mittelwert 0** und **Standardabweichung 1** pro Spalte
- (scikit-learn — ein erster Vorgeschmack; später im Kurs noch viel mehr)

```python
from sklearn.preprocessing import StandardScaler
```

```python
scaler = StandardScaler()
scaled = scaler.fit_transform(people[["age", "income"]])
```

```python
pd.DataFrame(scaled, columns=["age", "income"]).round(2)
```


## Was die Skalierung bewirkt hat

- Jeder Wert gibt nun an: "wie viele Standardabweichungen vom Spaltenmittelwert"
- Negativ = unter dem Durchschnitt, positiv = über dem Durchschnitt
- Die beiden Spalten befinden sich nun auf der **gleichen Skala**
- Die **Form** der Verteilung bleibt unverändert — nur die Zahlen ändern sich
- (Alternative: `MinMaxScaler` skaliert in den Bereich 0–1)


## Warum Encoding?

- Modelle benötigen **Zahlen**, keine Textkategorien
- `"North" / "South" / "West"` muss zu numerischen Spalten werden
- Wir können die Wörter nicht einfach in ein Modell eingeben


## `get_dummies` anwenden (One-Hot Encoding)

- `pd.get_dummies()` erstellt **eine Spalte pro Kategorie**
- Jede Spalte ist `True`/`False`: gehört diese Zeile zu dieser Kategorie?
- Dies nennt sich **One-Hot Encoding**

```python
pd.get_dummies(people, columns=["region"])
```


## Ordinal vs. One-Hot

- **One-hot** wenn Kategorien **keine Reihenfolge** haben (Region: North/South/West)
- Eine einzelne **gerankte Zahl** nur wenn eine **echte Reihenfolge** vorhanden ist
  (Größe: klein < mittel < groß → 0, 1, 2)
- Eine gerankte Zahl für ungeordnete Daten zu verwenden erfindet eine falsche Reihenfolge — vermeiden Sie dies


## Achtung: Erst aufteilen, dann skalieren

- Später teilen wir die Daten in einen **Trainings**- und einen **Test**-Teil auf
- Den Scaler nur auf den **Trainingsdaten** fitten, dann auf die Testdaten anwenden
- Ein Fitten auf **allen** Daten lässt Test-Informationen in das Training "durchsickern"
- Den Train/Test-Split behandeln wir später in dieser Woche — nur ein kurzer Hinweis für jetzt


## Zusammenfassung

- Bereinigen → vorbereiten: Modelle benötigen Zahlen auf vergleichbaren Skalen
- **Skalieren** Sie numerische Spalten mit `StandardScaler` (Mittelwert 0, Standardabweichung 1)
- **Kodieren** Sie Kategorien mit `get_dummies` (One-Hot)
- Achten Sie auf **Leakage**: Passen Sie den Scaler erst nach dem Aufteilen der Testdaten an
- Dies ist die Brücke von der Datenbereinigung zum maschinellen Lernen
